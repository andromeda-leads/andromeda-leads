# Вебинары → папка «Обучение. Продукты»

Нажми ▶ слева от ячейки (или «Выполнить все»). Скрипт сам скачает 8 файлов
и положит их в твою папку на Диске. Нужно один раз разрешить доступ к Диску.


```
FOLDER_ID = '1j2tEhSZ1OKUdsmubyV26uonLxc6xv3ue'
FILES = {
    "Веб 1. Вводный (25.06.2026).webm": "https://github.com/andromeda-leads/andromeda-leads/releases/download/webinars-2026/1.25.06.2026.webm",
    "Веб 2. Сколково часть 1 (26.06.2026).webm": "https://github.com/andromeda-leads/andromeda-leads/releases/download/webinars-2026/2.1.26.06.2026.webm",
    "Веб 4. Финальная часть о Сколково (01.07.2026).webm": "https://github.com/andromeda-leads/andromeda-leads/releases/download/webinars-2026/4.01.07.2026.webm",
    "Веб 5. Об МТК (02.07.2026).webm": "https://github.com/andromeda-leads/andromeda-leads/releases/download/webinars-2026/5.02.07.2026.webm",
    "Веб 6. Гранты ФСИ/ЦПИИ/РФРИТ (03.07.2026).webm": "https://github.com/andromeda-leads/andromeda-leads/releases/download/webinars-2026/6.03.07.2026.webm",
    "Веб 7. Сомнения и вопросы по Сколково и грантам (07.07.2026).webm": "https://github.com/andromeda-leads/andromeda-leads/releases/download/webinars-2026/7.07.07.2026.webm",
    "Веб 8. О боте (17.07.2026).webm": "https://github.com/andromeda-leads/andromeda-leads/releases/download/webinars-2026/8.17.07.2026.webm"
}

from google.colab import auth
auth.authenticate_user()
import google.auth, urllib.request, os
from googleapiclient.discovery import build
from googleapiclient.http import MediaFileUpload

creds, _ = google.auth.default()
svc = build('drive', 'v3', credentials=creds)

for name, url in FILES.items():
    path = '/content/' + name
    print('Скачиваю:', name, flush=True)
    urllib.request.urlretrieve(url, path)
    print('  загружаю в Диск...', flush=True)
    media = MediaFileUpload(path, mimetype='video/webm', resumable=True, chunksize=64*1024*1024)
    f = svc.files().create(body={'name': name, 'parents': [FOLDER_ID]},
                           media_body=media, fields='id,name,size').execute()
    print('  готово:', f['name'], f.get('size'), flush=True)
    os.remove(path)

print('ВСЁ ЗАГРУЖЕНО — открой папку «Обучение. Продукты»', flush=True)
```
